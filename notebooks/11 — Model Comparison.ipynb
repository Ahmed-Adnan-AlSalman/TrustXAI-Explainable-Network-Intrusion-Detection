```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("Libraries Loaded Successfully.")
```

```python
results = pd.DataFrame({

    "Model":[
        "Random Forest",
        "XGBoost"
    ],

    "Accuracy":[
        0.9900,
        0.9918
    ],

    "Precision":[
        0.9901,
        0.9919
    ],

    "Recall":[
        0.9900,
        0.9918
    ],

    "F1 Score":[
        0.9900,
        0.9918
    ]

})

results
```

```python
results.to_csv(
    r"D:\TrustXAI\results\Model_Comparison.csv",
    index=False
)

print("Comparison Table Saved Successfully.")
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["Accuracy"]
)

plt.ylim(0.98,1.00)

plt.title("Accuracy Comparison")

plt.ylabel("Accuracy")

plt.tight_layout()

plt.savefig(
    r"D:\TrustXAI\results\Accuracy_Comparison.png",
    dpi=300
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["Precision"]
)

plt.ylim(0.98,1.00)

plt.title("Precision Comparison")

plt.tight_layout()

plt.savefig(
    r"D:\TrustXAI\results\Precision_Comparison.png",
    dpi=300
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["Recall"]
)

plt.ylim(0.98,1.00)

plt.title("Recall Comparison")

plt.tight_layout()

plt.savefig(
    r"D:\TrustXAI\results\Recall_Comparison.png",
    dpi=300
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    results["Model"],
    results["F1 Score"]
)

plt.ylim(0.98,1.00)

plt.title("F1 Score Comparison")

plt.tight_layout()

plt.savefig(
    r"D:\TrustXAI\results\F1_Comparison.png",
    dpi=300
)

plt.show()
```
