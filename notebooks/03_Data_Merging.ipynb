```python
import os
import glob
import pandas as pd

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# =====================================
# Project Paths
# =====================================

PROJECT_ROOT = r"D:\TrustXAI"

PROCESSED_DATA_PATH = os.path.join(PROJECT_ROOT, "data", "processed")
FINAL_DATA_PATH = os.path.join(PROJECT_ROOT, "data", "final")

os.makedirs(FINAL_DATA_PATH, exist_ok=True)

print("Processed Path :", PROCESSED_DATA_PATH)
print("Final Path     :", FINAL_DATA_PATH)
```

```python
processed_files = sorted(
    glob.glob(os.path.join(PROCESSED_DATA_PATH, "*.csv"))
)

print(f"Total cleaned files: {len(processed_files)}\n")

for i, file in enumerate(processed_files, 1):
    print(f"{i}. {os.path.basename(file)}")
```

```python
# ==========================================
# Merge all cleaned CSV files
# ==========================================

output_file = os.path.join(FINAL_DATA_PATH, "TrustXAI_Final_Dataset.csv")

first_file = True
total_rows = 0

for i, file in enumerate(processed_files, start=1):

    print("=" * 70)
    print(f"[{i}/{len(processed_files)}] Merging: {os.path.basename(file)}")

    df = pd.read_csv(file, low_memory=False)

    total_rows += len(df)

    df.to_csv(
        output_file,
        mode="w" if first_file else "a",
        header=first_file,
        index=False
    )

    first_file = False

    print(f"Rows Added : {len(df):,}")

print("\nMerge Completed Successfully!")
print(f"Total Rows : {total_rows:,}")
print(f"Saved To   : {output_file}")
```
