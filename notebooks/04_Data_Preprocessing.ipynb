```python
# =====================================
# Create Ready Dataset Folder
# =====================================

READY_DATA_PATH = os.path.join(PROJECT_ROOT, "data", "ready")
os.makedirs(READY_DATA_PATH, exist_ok=True)

print(READY_DATA_PATH)
```

```python
# =====================================
# Encode Label Column (Chunk Processing)
# =====================================

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

# قراءة جميع الفئات فقط
labels = pd.read_csv(
    FINAL_FILE,
    usecols=["Label"]
)["Label"]

encoder.fit(labels)

print("Classes Found:")
print(list(encoder.classes_))

output_file = os.path.join(
    READY_DATA_PATH,
    "TrustXAI_Encoded.csv"
)

CHUNK_SIZE = 100000
first_chunk = True

for chunk in pd.read_csv(
    FINAL_FILE,
    chunksize=CHUNK_SIZE,
    low_memory=False
):

    chunk["Label"] = encoder.transform(chunk["Label"])

    chunk.to_csv(
        output_file,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    first_chunk = False

print("\nLabel Encoding Completed Successfully!")
print("Saved To:", output_file)
```
