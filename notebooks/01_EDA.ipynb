```python
# ==========================
# Project: TrustXAI
# Notebook: 01_EDA
# ==========================

import os
import glob
import pandas as pd

# Path to raw data
data_path = "../data/raw/*.csv"

# Read all CSV files
files = glob.glob(data_path)

print("Number of CSV files:", len(files))

# Load and combine all datasets
df_list = []

for file in files:
    print("Loading:", os.path.basename(file))
    temp = pd.read_csv(file, encoding="latin1", low_memory=False)
    df_list.append(temp)

# Merge all datasets
df = pd.concat(df_list, ignore_index=True)

print("\nDataset loaded successfully!")
```
