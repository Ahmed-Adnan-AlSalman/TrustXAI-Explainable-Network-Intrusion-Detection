```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("Libraries Loaded Successfully.")
```

```python
comparison = pd.read_csv(
    r"D:\TrustXAI\results\Model_Comparison.csv"
)

comparison
```

```python
best_model = comparison.loc[
    comparison["Accuracy"].idxmax()
]

print("="*50)
print("Best Model")
print("="*50)

print(best_model)
```

```python
final_table = comparison.copy()

final_table = final_table.round(4)

final_table.to_csv(
    r"D:\TrustXAI\results\Final_Results_Table.csv",
    index=False
)

final_table
```

```python
report = f"""

Final Results Summary

Best Model:
{best_model['Model']}

Accuracy:
{best_model['Accuracy']:.4f}

Precision:
{best_model['Precision']:.4f}

Recall:
{best_model['Recall']:.4f}

F1 Score:
{best_model['F1 Score']:.4f}

Conclusion:

XGBoost achieved the highest overall performance among all evaluated models.

"""

print(report)

with open(
    r"D:\TrustXAI\results\Final_Report.txt",
    "w"
) as f:

    f.write(report)
```

```python
comparison.set_index("Model").plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Final Performance Comparison")

plt.ylabel("Score")

plt.xticks(rotation=0)

plt.ylim(0.98,1.00)

plt.tight_layout()

plt.savefig(
    r"D:\TrustXAI\results\Final_Performance.png",
    dpi=300
)

plt.show()
```
